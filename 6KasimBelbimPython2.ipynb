# any, all operatörleri
`any(iterable)`
Herhangi bir değer true ise True döndürür. Eğer iterable nesne tamamen boş ise False döndürür.

`all(iterable)`
Tüm değerler  true ise True döndürür. Iterable nesne tamamen boş ise yine True döndürür.


```python
any([False, False, 0, 0.0, '']) # Herhangi biri True mu
```

```python
any([False, False, 0, 0.0, '', 0.1])
```

```python
all(["False", "False", "0", "0.0", '...',8]) # Hepsi True mu?
```

```python
all(2, 5, 'aslı', 5.6)  # tek bir grup içerisine almamızı istiyor
```

```python
all((2, 5, 'aslı', 5.6))
```

```python
# any ve all gruplama ister
any([False, False, 0, 0.0, ''])
```

```python
all(("False", "False", "0", "0.0", '...',8))
```

```python
# Bir firmanın yaşa göre uyguladığı inidirimi hesaplayan program
Fiyat = 100
yas = int (input('Yaşınız.:'))
print ("Normal Bilet Fiyatı.:", Fiyat)
if yas>0 and yas<=6:
   iFyt = 0                # Ücretsiz
elif yas>=7 and yas<=12:
   iFyt = Fiyat/2          # %50 indirim
elif yas>=13 and yas<=26:
   iFyt = Fiyat-Fiyat*0.20 # %20 indirim
elif yas>=27 and yas<=59:
   iFyt = Fiyat            # indirim yok
elif yas>=60 and yas<=64:
   iFyt = Fiyat-Fiyat*0.20 # %20 indirim
elif yas>=65:
   iFyt = Fiyat/2          # %50 indirim
else:
   print("Hatalı yaş değeri girildi...")

print ("İndirimli Bilet Fiyatı.:", iFyt)
```
