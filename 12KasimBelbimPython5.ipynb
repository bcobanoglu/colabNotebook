# DOSYALAMA(Files)

```python
# Girilen bir ismi 5 kez alt alta dosyaya yazalım
DOSYA = open("ad.txt", "w")
ad = input("Bir isim gir..:")
for i in range(5):
    DOSYA.write(ad + "\n")
DOSYA.close()
```

```python
# Aynı işi print() fonksiyonu ile yazabilir miyiz?
DOSYA = open("ad2.txt", "w")
ad = input("Bir isim gir..:")
print((ad + "\n")*5, file=DOSYA)
DOSYA.close()
```

```python
# Dosya okuma işlemi
DOSYA = open("ad.txt", "r")
for satir in DOSYA:
    print(satir)
DOSYA.close()
```

```python
# with open ile
with open("ad.txt", "r") as DOSYA:
    data = DOSYA.read()

# data' da tutulan içeriği ekranda göster
print(data)
```

# JSON Dosyaları Okuma


JSON (JavaScript Object Notation / JavaScript Nesne Gösterimi), farklı programlama dilleri arasında yapılandırılmış veri alışverişi yapmayı sağlayan, hafif ve metin tabanlı bir veri formatıdır.

```python
json_data = {
  "ad": "Ahmet",
  "yas": 30,
  "ogrenciMi": False,
  "yetenekler": ["Yazılım", "Tasarım", "Analiz"],
  "adres": {
    "sehir": "İstanbul",
    "ulke": "Türkiye"
  }
}
```

```python
# bu datayı json file nasıl yazarım?
# 1.yöntem json modülünü kullanmak
import json
# encoding parametresi open() fonksiyonu içinde tanımlanmalıdır
with open("personel.json", mode="w", encoding="utf-8") as DOSYA:
    # ensure_ascii=False Türkçe karakterlerin düzgün görünmesini sağlar
    json.dump(json_data, DOSYA, indent=2, ensure_ascii=False)
```

```python
# json file okuma
import json
with open("personel.json", mode="r", encoding="utf-8") as DOSYA:
    data = json.load(DOSYA)

print(data)
```

```python
# json file okuma
import json
with open("personel.json", mode="r", encoding="utf-8") as DOSYA:
    data = json.load(DOSYA)

# bir döngü içerisinde daha güzel formatta yazdırabiliriz.
for k,v in data.items():
    print(k, ":", v)
```

```python
# 2.yöntem pandas ile okuma (Hata çözümü)
import pandas as pd
import json

# İç içe geçmiş (nested) JSON verilerini düzleştirmek için json_normalize kullanırız:
with open("personel.json", mode="r", encoding="utf-8") as DOSYA:
    data = json.load(DOSYA)

df = pd.json_normalize(data)
df
```
