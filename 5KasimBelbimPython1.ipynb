#escape karakterler

**Escape (kaçış) karakterleri**, metin içinde özel bir anlam taşıyan karakterlerdir. Ters bölü (**``\``** , *backslash*) ile başlar ve ardından gelen karakterin normal anlamının dışında yorumlanmasını sağlar.

| Escape | Anlamı |
|---|---|
| `\n` | Yeni satır (newline) |
| `\t` | Yatay sekme (tab) |
| `\\` | Ters bölü karakterinin kendisi |
| `\'` | Tek tırnak |
| `\"` | Çift tırnak |
| `\r` | Satır başına dönüş (carriage return) |
| `\b` | Geri silme (backspace) |
| `\a` | Sistem zili (bell) |
| `\0` | Null karakter |
| `\xhh` | 16'lık (hex) kodlu karakter |
| `\uXXXX` | 16 bitlik Unicode karakter |
| `\N{isim}` | Unicode ismi ile karakter |

## Docstring
Docstring'ler bir modül, fonksiyon ya da programın işlevini açıklayan dokumanlardır.

Docstring metinlerine `__doc__` veya `help()` deyimleri ile erilşebiliriz.

docstring metinleri ''' veya """ ile oluşturulur. Örnek;


```
def merhaba(ad="Bulent"):

    """
    Bu selam veren bir fonksiyon
    """
    print(ad + " Hoca")
```


doxstring ler ayrıca test yazılımlarında kullanılır.






````python
help(print)
````

````python
print(print.__doc__)
````
