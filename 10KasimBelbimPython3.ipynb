# 1. match - case
Python **3.10** ile gelen bir **desen eşleştirme** yapısıdır. Çok sayıda `if / elif` yerine kullanılabilir.
Sürümünüzü kontrol edin:

````python
import sys
sys.version
````

Bir web sitesine girişte karşılaşılabilecek HTTP istemci hata kodlarını gösteren bir program;

````python
status = int(input("Web page status number..:"))
match status:
   case 400:
       print("Bad request")
   case 404:
       print("Not found")
   case 401 | 403:  # 401 veya 403 ise
       print("Not allowed")
   case 418:
       print("I'm a teapot")
   case s if 200 <= status <= 299:
       print(f"{s}: Success")
   case _:  # else
       print("Something's wrong with the internet")
````

````python
# aynı sözlük yapısı farklı şekillerde tasarlanabilir
a = dict(one=1, two=2, three=3)
b = {'one': 1, 'two': 2, 'three': 3}
c = dict(zip(['one', 'two', 'three'], [1, 2, 3]))
d = dict([('two', 2), ('one', 1), ('three', 3)])
e = dict({'three': 3, 'one': 1, 'two': 2})
f = dict({'one': 1, 'three': 3}, two=2)
a == b == c == d == e == f
````

````python
sozluk2 = dict(one=1, two=2, three=3)
print(sozluk2)
ters_sozluk = dict(reversed(sozluk2.items()))
print(ters_sozluk)
````

````python
list(sozluk2.keys())
````

````python
d = {"one": 1, "two": 2, "three": 3, "four": 4}
print(d)

print(list(d)) # sadece key'ler listeye dönüşür

print(list(d.values())) #value'lar listeye dönüşür

d["one"] = 42
print(d)

del d["two"]
d["two"] = None
print(d)
````
