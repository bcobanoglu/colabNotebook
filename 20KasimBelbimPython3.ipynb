````python
import random

print(random.random())          # 0.0 ile 1.0 arası float
print(random.randint(1, 10))    # 1 ve 10 dahil tam sayı
print(random.uniform(1.5, 5.5)) # aralıkta float değer
print(random.randrange(0, 10, 2))  # 0,2,4,6,8'den biri
````

````python
# password üreteci tasarlayalım
import string
import random
k_harf = random.sample(string.ascii_lowercase, 2)
b_harf = random.sample(string.ascii_uppercase, 2)
rakam = random.sample(string.digits, 2)
ozel = random.sample(string.punctuation, 2)
password = k_harf + b_harf + rakam + ozel
random.shuffle(password)
password = "".join(password)
print("Üretilen Şİfre..:", password)

````

````python
# Sayı tahmin oyunu:
import random
tutulan = random.randint(1,100)
sayac = 1
puan = 100
while True:
    tahmin = int(input("Tahmininiz..: "))
    sayac += 1
    if tahmin == tutulan:
        print(f"Tebrikler, {sayac}. defada bildiniz")
        break
    elif tahmin<tutulan:
        print("Daha büyük bir sayı deneyiniz")
    else:
        print("Daha küçük bir sayı deneyiniz")

    puan = puan-10

print("Puanınız..:", puan)
````

## `zip()` fonksiyonu

Birden fazla iterable'ı eleman eleman eşleştirir.
Sonuç bir **iterator**'dır; görmek için `list()` veya `dict` sarmalına almak gerekir.

````python
isimler = ["Ali", "Ayşe", "Can"]
yaslar = [20, 22, 21]
sehirler = ["Sakarya", "İzmir", "Ankara"]

print(list(zip(isimler, yaslar)))
print(list(zip(isimler, yaslar, sehirler)))

# Dict oluşturma
print(dict(zip(isimler, yaslar)))
````

````python
# Bir sözlük tasarlayalım. Sonrada bu sözlük üzerinde
# arama , slime ve güncelleme işlemlerini yapalım
my_dict = {'one': 1,
           'two': 2,
           'three': 3,
           'four': 4,
           'five':5}
# Sözlükte arama işlemi
aranan = input("İngilizce Kelime..: ")
print(my_dict.get(aranan, "Sözlükte bu kelime yoktur"))
````

````python
# setdefault ile arama işlemi
aranan = input("İngilizce Kelime..: ")
print(my_dict.setdefault(aranan, "Sözlüğe eklendi"))
````
