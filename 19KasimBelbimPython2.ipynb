# any, all operatörleri
`any(iterable)`
Herhangi bir değer true ise True döndürür. Eğer iterable nesne tamamen boş ise False döndürür.

`all(iterable)`
Tüm değerler  true ise True döndürür. Iterable nesne tamamen boş ise yine True döndürür.


````python
any([False, False, 0, 0.0, '']) # Herhangi biri True mu
````

````python
any([False, False, 0, 0.0, '', 0.1])
````

````python
False or False or 0 or 0.0 or '' or 0.1
````

````python
False or True or 0 or 0.0 or '' or 0.1
````

````python
all(["False", "False", "0", "0.0", '...',8]) # Hepsi True mu?
````

````python
all(2, 5, 'aslı', 5.6)  # tek bir grup içerisine almamızı istiyor
````

````python
all((2, 5, 'aslı', 5.6))
````

````python
# any ve all gruplama ister
any([False, False, 0, 0.0, ''])
````

````python
all(("False", "False", "0", "0.0", '',8))
````

````python
all("") # içi boş bir iterable da TRUE
````

````python
any("") # içi boş bir iterable da FALSE döndürür
````

````python
any("000000000") # içi dolu bir string gibi
````

````python
all("000000000") # içi dolu bir string gibi
````

#match-case yapısı

Diğer dillerdeki `switch`'ten çok daha güçlü bir **yapısal desen eşleştirme** (structural pattern matching) özelliğidir.

```python
match değişken:
    case desen1:
        ...
    case desen2 | desen3:
        ...
    case _:          # varsayılan (wildcard)
        ...
```

````python
status = int(input("HTTP Code Numbers..:"))
match status:
    case _ if status>=200 and status<300:
        print("Başarılı")
    case 400:
        print("Hatalı İstek")
    case 404:
        print("Not found")
    case 401 | 403: # veya
        print("İzinsiz girişim")
    case 406:
        print("Not Acceptable")
    case _:
        print("Hatalı bir numara girişi")
````

````python
# gün girilsin, hafta içi mi? hafta sonu mu? bize söylesin..
gun = input("Gün giriniz..:").lower()
match gun:
    case "pazartesi"|"salı"|"çarşamba"|"perşembe"|"cuma":
        print("Hafta içi")
    case "cumartesi"|"pazar":
        print("Hafta Sonu")
    case _ :
        print("Geçersiz gün ismi...")
````
